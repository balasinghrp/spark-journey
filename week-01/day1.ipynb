```python
from pyspark.sql import SparkSession

spark = (SparkSession.builder
         .master("local[*]")
         .appName("day1")
         .getOrCreate())
spark
```

```python
from pyspark.sql import functions as F

df = spark.range(0, 1_000_000).withColumn("even", F.col("id") % 2 == 0)
df.show(5)
df.groupBy("even").count().show()
```
